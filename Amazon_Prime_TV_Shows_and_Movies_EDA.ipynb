### What all manipulations have you done and insights you found?

Data Manipulations

1. **Removed duplicate records**
   - Identified and removed 3 duplicate records from the `titles` dataset.

2. **Handled missing values in the `description` column**
   - Replaced missing descriptions with `"No Description"` instead of deleting the corresponding records or using mode imputation, as the column contains textual information.

3. **Handled missing values in the `age_certification` column**
   - Replaced missing values with `"Unknown"` because `age_certification` is a categorical variable and the missing values do not indicate any specific certification.

4. **Converted `genres` and `production_countries` to list format**
   - Converted these columns from string representations of lists into actual Python lists.
   - This allowed empty lists (`[]`) to be identified separately from missing (`NaN`) values and made the columns easier to analyze.

---

Key Insights

1. **`seasons` column**
   - A high percentage of values in the `seasons` column are missing.
   - Further investigation showed that missing values occur only for `MOVIE` records, while `SHOW` records contain season information.
   - Therefore, the missing values are structural rather than random.
   - The column was retained because `seasons` is applicable only to TV shows.

2. **`genres` and `production_countries` columns**
   - Initially, these columns contained list-like values stored as strings.
   - After converting them into actual lists, empty lists (`[]`) could be identified as records where genre or production-country information is unavailable.
   - These empty lists were not treated as `NaN` values because they represent a different type of missing information.

3. **`runtime` column**
   - Three records were identified with unusually high runtimes: 549, 540, and 328 minutes.
   - These values were treated as potential outliers and investigated individually.
   - The records were retained because an extreme value is not necessarily an incorrect value.

4. **Rating and popularity variables**
   - `imdb_score`, `imdb_votes`, `tmdb_popularity`, and `tmdb_score` contain some missing values.
   - These missing values were retained as `NaN` because missing information does not imply a value of zero.

````python
# Chart - 9 visualization code
# Distribution of TMDB ratings
sns.histplot(
    data=movies_df,
    x='tmdb_score',
    bins=20
)

plt.title('Distribution of TMDB Scores', fontweight='bold')
plt.xlabel('TMDB Score')
plt.ylabel('Number of Titles')

plt.show()
````

````python
# Analyze production_countries
countries_df = movies_df.explode('production_countries')

# countries_df['production_countries'].value_counts().head(10)

top_countries = countries_df['production_countries'].value_counts().head(10)

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Production Countries', fontweight='bold')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.show()
````

````python
# production country vs imdb score

top_country_names = (
    countries_df['production_countries']
    .value_counts()
    .head(10)
    .index
)

top_country_df = countries_df[
    countries_df['production_countries'].isin(top_country_names)
]

sns.boxplot(
    data=top_country_df,
    x='production_countries',
    y='imdb_score'
)

plt.title('IMDb Score Distribution by Production Country', fontweight='bold')
plt.xlabel('Production Country')
plt.ylabel('IMDb Score')

plt.xticks(rotation=45)

plt.show()
````

````python
# production country vs content type

country_type = pd.crosstab(
    top_country_df['production_countries'],
    top_country_df['type']
)

country_type.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Movies and Shows by Production Country', fontweight='bold')
plt.xlabel('Production Country')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)

plt.show()
````

````python
# Production Country × Average IMDb Score

country_rating = (
    top_country_df
    .groupby('production_countries')['imdb_score']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

sns.barplot(
    data=country_rating,
    x='imdb_score',
    y='production_countries'
)

plt.title('Average IMDb Score by Production Country', fontweight='bold')
plt.xlabel('Average IMDb Score')
plt.ylabel('Production Country')

plt.show()
````
