### Check Unique Values for each variable.

````python
# Check Unique Values for each variable.
unique_values = movies_df.nunique().reset_index()
unique_values.columns = ['Column', 'Unique Values']
unique_values
````

````python
# Check the unique values in selected columns bcz remaining columns has huge unique values.
temp_df = ['type', 'age_certification', 'seasons']

for column in temp_df:
    unique_values = movies_df[column].unique()
    print(f"{column}: {unique_values}")
````

````python
# Examine the distribution of age certifications, including missing values.
movies_df['age_certification'].value_counts(dropna=False)
````

````python
(movies_df['genres'] == '[]').sum()
````

````python
(movies_df['production_countries'] == '[]').sum()
````

````python
movies_df['seasons'].isna().sum()
````

````python
# Analyze missing seasons by content type to determine whether the missing values are expected.
movies_df.groupby('type')['seasons'].apply(lambda x: x.isnull().mean() * 100)
````

````python
movies_df['imdb_score'].value_counts(dropna=False)
````

````python
# Compare missing IMDb scores between movies and shows.
movies_df.groupby('type')['imdb_score'].apply(
    lambda x: x.isnull().mean() * 100
)
````

````python
# Compare missing IMDb scores between movies and shows.
movies_df.groupby('type')['imdb_votes'].apply(
    lambda x: x.isnull().mean() * 100
)
````

````python
# Check whether IMDb score and vote information are missing together.
movies_df[['imdb_score', 'imdb_votes']].isnull().sum()
````

````python
# Identify titles with unusually long runtimes.
movies_df[movies_df['runtime'] > 300][
    ['title', 'type', 'runtime']
].sort_values('runtime', ascending=False)
````

````python
# Check for empty genre and country lists.
print("Empty genres:", movies_df['genres'].apply(len).eq(0).sum())
print("Empty countries:", movies_df['production_countries'].apply(len).eq(0).sum())
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Delete duplicate values
movies_df.drop_duplicates(inplace = True)
````

````python
# Handle empty values in the production_countries column
movies_df['production_countries'] = movies_df['production_countries'].replace('[]', 'Unknown')
````

````python
movies_df['description'] = (
    movies_df['description'].fillna('No Description')
)
````

````python
movies_df['age_certification'] = (
    movies_df['age_certification'].fillna('Unknown')
)
````

````python
import ast

movies_df['genres'] = movies_df['genres'].apply(ast.literal_eval)
````
